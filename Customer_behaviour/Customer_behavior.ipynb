```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns=df. columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
df.columns
```

```python
##Create a new column -Age-Group
```

```python
labels=['Young Adult', 'Adult ', ' Middle-Aged ', 'Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
df[['age','age_group']].head(10)
```

```python
###Purchase frequency date

frequency_days = {
    'Weekly': 7,
    'Bi-Weekly': 14,
    'Fortnightly': 14,
    'Monthly': 30,
    'Quarterly': 90,
    'Annually': 365
}

df['purchasing_frequency_days'] = df['frequency_of_purchases'].map(frequency_days)
df.head()
```

```python
###lets check wheether discount_applied	& promo_code_used coloumns are same means if we use promocode then discount will be the there if same the then we remove the column
(df['discount_applied']==df['promo_code_used']).all()
```

```python
df=df.drop('promo_code_used',axis=1)
```
